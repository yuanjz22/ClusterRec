```python
from datasets import load_dataset
import os
import random
import numpy as np

fname = 'Industrial_and_Scientific'
local_path = f"./data_{fname}/"
    
# 1. 加载元数据 (两个 Parquet 文件)
print("Loading local Meta Data...")
meta_files = [
    os.path.join(local_path, "full-00000-of-00002.parquet"), # 确认一下你本地的实际文件名
    os.path.join(local_path, "full-00001-of-00002.parquet")
]
# 使用 'parquet' 读取器直接加载本地文件
meta_dataset = load_dataset("parquet", data_files={"full": meta_files})
```

```python
# 定义三个本地文件的路径
data_files = {
    "train": os.path.join(local_path, f"{fname}.train.csv"),
    "valid": os.path.join(local_path, f"{fname}.valid.csv"),
    "test": os.path.join(local_path, f"{fname}.test.csv")
}

print("Loading local 5core CSV split files...")
# 使用 'csv' 加载器，指定 data_files 字典
# 这样加载出来的 dataset 对象自带 ['train'], ['valid'], ['test'] 三个 split
dataset = load_dataset("csv", data_files=data_files)
```
